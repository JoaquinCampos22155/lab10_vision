```python
import torch
import time
import gc
import pandas as pd
import matplotlib.pyplot as plt

from diffusers import StableDiffusionPipeline, AutoPipelineForText2Image
from PIL import Image
```

```python
print("CUDA disponible:", torch.cuda.is_available())

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))

prompt = "A highly detailed cinematic and futuristic fruit glowing in a cyberpunk laboratory, neon lights, 4k resolution"

seed = 42
device = "cuda"
```

```python
def benchmark_pipeline(pipe, steps, prompt, seed):

    torch.cuda.empty_cache()
    gc.collect()

    torch.cuda.reset_peak_memory_stats()

    generator = torch.manual_seed(seed)

    start_time = time.time()

    result = pipe(
        prompt,
        num_inference_steps=steps,
        generator=generator
    )

    end_time = time.time()

    elapsed_time = end_time - start_time

    max_memory = torch.cuda.max_memory_allocated() / (1024 ** 3)

    image = result.images[0]

    return {
        "image": image,
        "time": elapsed_time,
        "memory": max_memory
    }
```

```python
#Stable Diffusion
pipe_a = StableDiffusionPipeline.from_pretrained(
    "runwayml/stable-diffusion-v1-5",
    torch_dtype=torch.float16
).to(device)
pipe_a.safety_checker = None
metrics_a = benchmark_pipeline(
    pipe=pipe_a,
    steps=50,
    prompt=prompt,
    seed=seed
)

#Guardamos imagen
image_a = metrics_a["image"]
#se libera memoria
del pipe_a
torch.cuda.empty_cache()
gc.collect()
```

```python
# Opcion B - SDturbo
pipe_b = AutoPipelineForText2Image.from_pretrained(
    "stabilityai/sd-turbo",
    torch_dtype=torch.float16,
    variant="fp16"
).to(device)
metrics_b = benchmark_pipeline(
    pipe=pipe_b,
    steps=4,
    prompt=prompt,
    seed=seed
)
image_b = metrics_b["image"]
```

```python
#copmaracion visual
fig, axes = plt.subplots(1, 2, figsize=(14, 7))

axes[0].imshow(image_a)
axes[0].set_title("Stable Diffusion 1.5 — 50 pasos")
axes[0].axis("off")

axes[1].imshow(image_b)
axes[1].set_title("SD-Turbo — 4 pasos")
axes[1].axis("off")


plt.tight_layout()
plt.show()
```

El modelo Stable Diffusion 1.5 con 50 pasos logró una reconstrucción mucho más rica en detalle y coherencia espacial, incluso podria parecer una especie de cocina futuristica cuando sabemos que el prompt busca un laboratorio y tal vez por eso mete esos "tubos" en el centro. La escena tiene profundidad, múltiples objetos bien definidos y sobre todo la iluminación neón más compleja. También se observa mejor resolución de altas frecuencias: reflejos, bordes, texturas metálicas y separación entre foreground/background. El modelo tuvo suficiente número de iteraciones para refinar progresivamente la estructura latente y converger correctamente.

Por otro lado, SD-Turbo con 4 pasos mantiene la semántica general del prompt, ya que todavía reconoce correctamente el entorno cyberpunk, las luces neón y la fruta brillante en el centro. Sin embargo, la imagen pierde mucho refinamiento fino. Las texturas están suavizadas, los contornos son menos precisos y aparecen artefactos visuales típicos de una convergencia acelerada, incluso parecen hechos en base de circulos o grumos. La iluminación también se vuelve más agresiva y saturada.

Lo interesante es que, aun con solo 4 pasos, SD-Turbo sigue generando una imagen coherente y reconocible. Por ejemplo las cosas verdes podemos reconoecrlas como peras o frutas y  el fondo el lab con los colores neon. Eso demuestra el efecto de la destilación: el modelo aprendió a aproximar la trayectoria completa de denoising en muy pocas iteraciones. Si el modelo estándar se ejecutara con únicamente 4 pasos, probablemente el resultado sería ruido residual o una escena incompleta, porque no fue entrenado para converger tan rápido.

```python
#Tabla comparativa
results_df = pd.DataFrame({
    "Modelo": [
        "Stable Diffusion 1.5",
        "SD-Turbo"
    ],
    "Pasos": [
        50,
        4
    ],
    "Tiempo (s)": [
        round(metrics_a["time"], 2),
        round(metrics_b["time"], 2)
    ],
    "VRAM (GB)": [
        round(metrics_a["memory"], 2),
        round(metrics_b["memory"], 2)
    ]
})

results_df
```

```python
# Gráfica — Tiempo de Inferencia

plt.figure(figsize=(6,4))

plt.bar(
    results_df["Modelo"],
    results_df["Tiempo (s)"]
)

plt.ylabel("Segundos")
plt.title("Tiempo de Inferencia")

plt.show()
```

```python
# Gráfica — Consumo de VRAM

plt.figure(figsize=(6,4))

plt.bar(
    results_df["Modelo"],
    results_df["VRAM (GB)"]
)

plt.ylabel("GB")
plt.title("Consumo Máximo de VRAM")

plt.show()
```

```python
# Gráfica — Pasos de Inferencia

plt.figure(figsize=(6,4))

plt.bar(
    results_df["Modelo"],
    results_df["Pasos"]
)

plt.ylabel("Número de pasos")
plt.title("Comparación de Pasos de Inferencia")

plt.show()
```

# RESULTADOS Y ANÁLISIS
Los resultados muestran el trade-off entre calidad visual y eficiencia computacional. Stable Diffusion 1.5 generó una imagen con más detalle en texturas, iluminación y estructura espacial. Esto ocurre porque el modelo realiza denoising progresivo durante 50 iteraciones, permitiendo refinar la representación latente antes del decoding final con el VAE.

SD-Turbo generó una imagen coherente utilizando únicamente 4 pasos. La calidad visual es menor y aparecen artefactos, pero la estructura semántica principal del prompt se mantiene. Esto refleja el efecto de la destilación: el modelo aprende una aproximación comprimida de la trayectoria de difusión, eliminando más ruido por iteración.

Las métricas reflejan este comportamiento. El tiempo de inferencia disminuyó fuertemente en SD-Turbo, reduciendo la latencia a una fracción del modelo estándar. El consumo de VRAM permaneció similar, indicando que el principal ahorro ocurre en tiempo computacional y no en memoria GPU. Esto sucede porque ambos pipelines siguen utilizando componentes pesados como la U-Net, el VAE y el text encoder.

Desde una perspectiva arquitectónica, SD-Turbo es una mejor opción para despliegue masivo debido a:
- menor latencia
- mayor throughput
- menor costo operativo por request
- mejor escalabilidad

Stable Diffusion 1.5 es más adecuado para escenarios donde la prioridad es maximizar fidelidad visual y detalle fino, como generación artística premium o rendering offline.

En conclusión, la destilación desplaza complejidad desde inferencia hacia entrenamiento, permitiendo aproximar el proceso completo de difusión en pocas iteraciones y haciendo viable el uso de modelos generativos en aplicaciones de gran escala y tiempo real.
