```python
import torch
from diffusers import StableDiffusionPipeline
import matplotlib.pyplot as plt
from PIL import Image
import numpy as np
```

```python
print("CUDA disponible:", torch.cuda.is_available())
if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
```

```python
model_id = "runwayml/stable-diffusion-v1-5"
device = "cuda"

prompt = "A highly detailed cinematic and futuristic fruit glowing in a cyberpunk laboratory, neon lights, 4k resolution"

num_steps = 20
seed = 42

capture_steps = [4, 10, 16, 20]
saved_latents = {}
```

modelo en GPU

```python
pipe = StableDiffusionPipeline.from_pretrained(
    model_id,
    torch_dtype=torch.float16
).to(device)

pipe.safety_checker = None
```

Fijar semilla

```python
generator = torch.manual_seed(seed)
```

# Callback para capturar latentes
Guardamos los tensores en pasos 4, 10, 16 y 20.

Ejecucion de de difusión

```python
def callback_fn(pipe, step, timestep, callback_kwargs):
    step_number = step + 1

    if step_number in capture_steps:
        latents = callback_kwargs["latents"]
        saved_latents[step_number] = latents.detach().clone()

    return callback_kwargs
_ = pipe(
    prompt=prompt,
    num_inference_steps=num_steps,
    generator=generator,
    callback_on_step_end=callback_fn,
)
```

## Decodificar latentes con VAE
aplicar scaling_factor antes de decodificar

```python
images = []

for step in capture_steps:
    latent = saved_latents[step]

    latent = latent / pipe.vae.config.scaling_factor

    with torch.no_grad():
        image = pipe.vae.decode(latent).sample

    image = (image / 2 + 0.5).clamp(0, 1)
    image = image.cpu().permute(0, 2, 3, 1).numpy()

    image = (image[0] * 255).astype(np.uint8)
    image = Image.fromarray(image)

    images.append((step, image))
```

## Visualizar

```python
fig, axes = plt.subplots(1, 4, figsize=(16, 4))

for ax, (step, img) in zip(axes, images):
    ax.imshow(img)
    ax.set_title(f"Step {step}")
    ax.axis("off")

plt.tight_layout()
plt.show()
```
